# Connecting Python to MySQL – Step‑by‑Step Guide
This notebook teaches you how to connect Python with MySQL using **mysql-connector-python**, create a database, create tables, and insert data.

We assume you already have **MySQL Workbench** installed and know your MySQL username and password.

## Step 1: Install MySQL Connector
Run the following command in your terminal (not inside the notebook):
```
pip install mysql-connector-python
```
This library allows Python to communicate with MySQL.

## Step 2: Connect to MySQL
Replace the password with your actual MySQL Workbench password.

````python
import mysql.connector

con = mysql.connector.connect(
    host="localhost",
    user="root",
    password="varun@123",
      database='school'  # replace with your MySQL Workbench password
)

if con.is_connected():
    print("Connected to MySQL successfully!")

con.close()
````

## Step 3: Create a Database in MySQL
This command creates a database named **college** if it does not already exist.

````python
con = mysql.connector.connect(
    host="localhost",
    user="root",
    password="your_password_here"
)
cur = con.cursor()

cur.execute("CREATE DATABASE IF NOT EXISTS college;")
print("Database created or already exists.")

con.close()
````

## Step 4: Connect to the `college` Database and Create a Table

````python
con = mysql.connector.connect(
    host="localhost",
    user="root",
    password="your_password_here",
    database="college"
)

cur = con.cursor()
cur.execute("""
CREATE TABLE IF NOT EXISTS students(
    id INT,
    name VARCHAR(50),
    marks INT
)
""")

con.commit()
print("Table created.")
con.close()
````

## Step 5: Insert Data into Table

````python
con = mysql.connector.connect(
    host="localhost",
    user="root",
    password="your_password_here",
    database="college"
)

cur = con.cursor()

cur.execute("INSERT INTO students VALUES (1, 'Varun', 95);")
con.commit()

print("Data inserted successfully!")
con.close()
````

## Step 6: Fetch and Display Data

````python
con = mysql.connector.connect(
    host="localhost",
    user="root",
    password="your_password_here",
    database="college"
)
cur = con.cursor()

cur.execute("SELECT * FROM students;")
rows = cur.fetchall()

for row in rows:
    print(row)

con.close()
````
