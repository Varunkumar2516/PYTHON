# Complete Guide to SQLite3 in Python

## 1. What is Database Programming?
**Database programming** means writing Python programs that can connect to a database, store data, update it, delete it, and retrieve it whenever needed. It is used when we want to keep data permanently instead of temporarily storing it in variables or files.

Python provides strong support for database programming through its built-in libraries and external modules. It can work with almost all types of relational databases such as SQLite, MySQL, PostgreSQL, Oracle, and SQL Server. Python follows a standard called **DB-API (Database Application Programmer’s Interface)** which defines how Python should communicate with any relational database.

### Why is it needed?
Database programming is needed because real-world applications such as banking systems, e-commerce websites, inventory systems, and login systems require safe and permanent storage of data. Using databases improves data security, consistency, accuracy, and reliability.

### Steps in Python Database Programming
In Python, a typical database program includes the following steps:
1. **Importing** the database module (e.g., `sqlite3` or `mysql.connector`)
2. **Creating a connection** using `connect()`
3. **Creating a cursor** to execute SQL commands
4. **Executing SQL queries** like `CREATE`, `INSERT`, `UPDATE`, `DELETE`, and `SELECT`
5. **Fetching results** using `fetchone()` or `fetchall()`
6. **Committing changes** using `commit()`
7. **Closing the connection** at the end

Python also supports prepared statements to prevent SQL injection and transaction management to ensure data is consistent even if failures occur. Because of its simplicity and wide library support, Python has become one of the most preferred languages for database application development.

```python
import sqlite3

# STEP 1 & 2: Create connection
# This creates the file 'my_database.db' if it does not exist
con = sqlite3.connect("my_database.db")

# STEP 3: Create a cursor object
# The cursor is used to traverse the records from the result set
cur = con.cursor()

print("Database connected successfully!")
```

```python
# STEP 4: Create a Table
# We use IF NOT EXISTS to prevent errors if we run the cell twice
cur.execute("""
CREATE TABLE IF NOT EXISTS students(
    id INTEGER PRIMARY KEY,
    name TEXT,
    grade TEXT,
    marks INTEGER
)
""")

print("Table created successfully!")
```

```python
# STEP 4 & 6: Insert Data (CREATE in CRUD)

# Method 1: Inserting a single row
cur.execute("INSERT INTO students (id, name, grade, marks) VALUES (1, 'Varun', '10th', 90)")

# Method 2: Inserting multiple rows using a list
students_list = [
    (2, 'Amit', '10th', 85),
    (3, 'Sneha', '11th', 92),
    (4, 'Rahul', '11th', 78)
]

# We use '?' as placeholders for security (Prepared Statements)
cur.executemany("INSERT INTO students VALUES (?, ?, ?, ?)", students_list)

# CRITICAL: Commit the transaction to save changes
con.commit()

print("Data inserted and saved!")
```

```python
# STEP 5: Fetch Data (READ in CRUD)

print("--- All Students ---")
cur.execute("SELECT * FROM students")
rows = cur.fetchall()

for row in rows:
    print(row)

print("\n--- Students with marks > 80 ---")
cur.execute("SELECT name, marks FROM students WHERE marks > 80")
toppers = cur.fetchall()
for student in toppers:
    print(student)
```

```python
# UPDATE Data (UPDATE in CRUD)
# Let's change Rahul's marks to 82

cur.execute("UPDATE students SET marks = 82 WHERE name = 'Rahul'")
con.commit()

print("Rahul's marks updated.")

# Verify update
cur.execute("SELECT * FROM students WHERE name='Rahul'")
print(cur.fetchone())
```

```python
# DELETE Data (DELETE in CRUD)
# Let's delete the student with ID 1

cur.execute("DELETE FROM students WHERE id = 1")
con.commit()

print("Student ID 1 deleted.")

# Verify deletion
cur.execute("SELECT * FROM students")
print(cur.fetchall())
```

```python
# STEP 7: Close Connection
con.close()
print("Connection closed.")
```
