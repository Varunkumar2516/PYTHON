**Regular Expressions (RegEx)** are patterns used to match text.

They are used for:
* Searching
* Validating
* Extracting
* Replacing text

**Common Use Cases:**
* Check if an email is valid
* Find all mobile numbers in text
* Validate passwords
* Search for specific words
* Replace unwanted characters

Python supports RegEx using the built-in `re` module.



> Regular expressions are patterns used for matching and manipulating text. They allow developers to define search patterns to validate, extract, or substitute parts of a string. Python provides the `re` module, which supports standard regex operations such as searching, matching, splitting, and replacing text. Regular expressions use special characters and symbols to define patterns, making them powerful for tasks like email validation, mobile number checking, password verification, and data cleaning. Because of their flexibility and speed, regular expressions are widely used in text processing, web scraping, data analysis, and form validation.

# 2. Python `re` Module — Basic Functions

```python
import re  # We must import this module first
```

### 🔷 1. re.match()
Checks if the pattern matches ONLY at the **beginning** of the string.

```python
result = re.match("Hello", "Hello Varun")
print(f"Match found: {result}")

# If it's not at the start, it returns None
result_fail = re.match("Varun", "Hello Varun")
print(f"Match failed: {result_fail}")
```

### 🔷 2. re.search()
Searches for the pattern **anywhere** in the string. It returns the first match found.

```python
result = re.search("Varun", "Hello Varun")
print(f"Search result: {result}")
```

### 🔷 3. re.findall()
Returns **all** non-overlapping matching strings in a list.

```python
# \d+ means "one or more digits"
result = re.findall(r"\d+", "My marks are 90 and 85")
print(result)
```

### 🔷 4. re.sub()
Used to **replace** text.

```python
text = "This is bad"
new_text = re.sub("bad", "good", text)
print(new_text)
```

### 🔷 5. re.split()
Splits the string based on the pattern.

```python
text = "a,b,c,d"
result = re.split(",", text)
print(result)
```

# 3. Special Symbols / Metacharacters (Deep Dive)

Here is the code demonstration for **every** important metacharacter.

### 1. `.` (Dot)
Matches **any** character except a newline.

```python
# Matches 'h', any char, then 't'
data='hello imvarun'
print(re.findall(r".", data))
```

### 2. `^` (Caret)
Matches the **start** of the string.

```python
text = "Hello World"
# Checks if text starts with Hello
if re.search(r"^Hello", text):
    print("Starts with Hello")
else:
    print("Does not start with Hello")
```

### 3. `$` (Dollar)
Matches the **end** of the string.

```python
text = "Hello World"
# Checks if text ends with World
if re.search(r"World$", text):
    print("Ends with World")
```

### 4. `*` (Star)
Matches **0 or more** occurrences of the preceding character.

```python
# 'ab*' matches 'a' followed by zero or more 'b's
# matches: 'a', 'ab', 'abb', 'abbb'

data='hello this is very intersting yes '
print(re.findall(r"l*", data))
```

### 5. `+` (Plus)
Matches **1 or more** occurrences of the preceding character.

```python
# 'ab+' matches 'a' followed by at least one 'b'
# matches: 'ab', 'abb' (Does NOT match 'a' or 'ac')
data='a aa aa aaaaaaaa aaaaaaaaaaa '
print(re.findall(r"a+", data))
```

### 6. `?` (Question Mark)
Matches **0 or 1** occurrence of the preceding character.

```python
# 'ab?' matches 'a' followed by zero or one 'b'
# matches: 'a', 'ab'
print(re.findall(r"ab?", "a ab abb ac"))
# b can comes or not (optional)
```

### 7. `{m}` and `{m,n}` (Braces)
Matches exact or range of repetitions.

```python
text = "a aa aaa aaaa aaaaa"

# Matches exactly 3 'a's
print("Exactly 3:", re.findall(r"a{3}", text))

# Matches between 2 and 4 'a's
print("2 to 4:", re.findall(r"a{2,4}", text))
```

### 8. `[]` (Square Brackets - Character Set)
Matches any one character inside the brackets.

```python
# [abc] matches a OR b OR c
print(re.findall(r"[abc]", "apple banana cherry"))

# [a-m] matches any lowercase letter from a to m
print(re.findall(r"[a-m]", "xyz abc"))
```

### 9. `[^ ]` (Negated Set)
Matches any character NOT inside the brackets.

```python
# [^a-z] matches anything that is NOT a lowercase letter
print(re.findall(r"[^a-z]", "HeLLo 123!"))
```

### 10. `\d` (Digit)
Matches any digit (0-9).

```python
text = "Room 101, Room 202"
print(re.findall(r"\d+", text))
```

### 11. `\w` (Word Character)
Matches alphanumeric characters (a-z, A-Z, 0-9, and underscore).

```python
text = "user_name123!"
# Matches words/identifiers, stops at !
print(re.findall(r"\w+", text))
```

### 12. `\s` (Whitespace)
Matches spaces, tabs, newlines.

```python
text = "Hello   World\nPython"
# Replaces all whitespace with a hyphen
print(re.sub(r"\s", "-", text))
```

### 13. `()` (Grouping)
Groups patterns together.

```python
# Matches 'ab' followed by 'c' 
# (ab)+ means 'ab' can repeat
print(re.findall(r"(ab)+c", "ababc abababc abc"))
```

# 4. Code Examples (Option B)

```python
# Example 1: Find all numbers
text = "Varun scored 88 marks in Math and 92 in Science."
nums = re.findall(r"\d+", text)
print("Numbers found:", nums)
```

```python
# Example 2: Validate Email
# Pattern explanation:
# ^        : Start of string
# [\w\.]+  : Word chars or dots (one or more)
# @        : Literal @ symbol
# [a-zA-Z]+: Letters (domain name)
# \.       : Literal dot
# [a-zA-Z]{2,3}$ : Extension (2 or 3 letters) at end of string

pattern = r"^[\w\.]+@[a-zA-Z]+\.[a-zA-Z]{2,3}$"
email = "varun123@gmail.com"

is_valid = bool(re.match(pattern, email))
print(f"Is '{email}' valid? {is_valid}")
```

```python
# Example 3: Check if string starts with "Hello"
print(bool(re.match(r"^Hello", "Hello World")))
```

```python
# Example 4: Replace digits with X
print(re.sub(r"\d", "X", "My number is 9876543210"))
```

### 5. Pattern Objects
You can compile a regex for repeated use, which improves performance.

```python
p = re.compile(r"\d+")
print(p.findall("Marks: 90 85 88 70"))
```
